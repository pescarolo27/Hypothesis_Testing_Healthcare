# Analysis I - Adverse Effects Between Drug/Placebo Groups
**Determine if the proportion of adverse effects differs significantly between the Drug & Placebo groups, saving the p-value as a variable called `two_sample_p_value`.**

This section will evaluate if there is a significant difference between the proportions of individuals across the drug (treatment) & placebo (control) groups who experienced adverse effects. To start, a brief, exploratory analysis can be performed to get an initial evaluation of these variables.

```python
#Calculate, inspect the ratios between 'trx' & 'adverse_effects'
trx_adv_eff_splits = orig_df.groupby(['trx','adverse_effects']).age.count().reset_index().rename(columns={'age':'#'})

#Want to add a variable indicating the percentage of individuals per 'trx' group
#retrieve the # individuals per 'trx' group
num_Drug_grp = orig_df[orig_df['trx'] == 'Drug'].shape[0]
num_Plac_grp = orig_df[orig_df['trx'] == 'Placebo'].shape[0]
#perform arithmetic; divide '#' by # individuals per 'trx' group, convert to a %
perc_Drug_grp = [np.round(100*i/num_Drug_grp,2) for i in trx_adv_eff_splits[trx_adv_eff_splits.trx == 'Drug']['#'].values]
perc_Plac_grp = [np.round(100*i/num_Plac_grp,2) for i in trx_adv_eff_splits[trx_adv_eff_splits.trx == 'Placebo']['#'].values]
#Create the new variable
trx_adv_eff_splits['% per trx group'] = perc_Drug_grp + perc_Plac_grp
#Check the results
display(trx_adv_eff_splits)


#Visualize the data
ax1 = sns.barplot(data=trx_adv_eff_splits, x='trx', y='% per trx group', hue='adverse_effects')
#edit plot details
plt.title('Proportions of Patients With/Without Adverse Effects per Group', fontsize=11)
plt.xlabel('Treatment Group')
plt.xticks(fontsize=9)
plt.ylabel('Proportion of Individuals per Group (%)', fontsize=9.5)
ax1.set(ylim=(0,100))
plt.yticks(ticks=range(0,101,10), fontsize=9)
plt.legend(title='Adverse Effects?', title_fontsize=9, fontsize=8, loc='upper right')
plt.grid(axis='y',alpha=0.2,color='gray')
plt.tight_layout();
```

The figure above illustrates the percentage of individuals who have experienced adverse effects from the drugs they took in the experimental trial based on what treatment group they were in.

Recall that in this trial, there were about two people in the "Drug" group for every one person in the "Placebo" group. Nonetheless, the proportions of individuals across the two treatment groups appear to be very similar. The proportion of people who experienced adverse effects was about 9.55% & 9.52% for those in the drug & placebo group respectively.  
As such, this figure suggests that there is no significant difference in the proportion of people who experienced adverse effects between the drug & placebo groups. Nevertheless, hypothesis testing can be utilized to evaluate this further.

Given the goal of this section, it would be appropriate to use a two-sample proportion z-test which tests for equality between proportions across two groups.
- Such a test requires the groups to be independent. Given that the pharmaceutical company performed a randomized, controlled trial where the patients were only part of either the "Drug" or "Placebo" group, it can be said that the two groups are independent.
- Each group needs to have a reasonably large sample size. The "Drug" & "Placebo" groups recorded data for about 10,700 & 5,400 individuals respectively which are sufficient.

In this case, the null & alternative hypotheses follow as:
- $H_0$: The proportion of people who experienced adverse effects was equivalent between the Drug & Placebo groups.
- $H_A$: The proportion of people who experienced adverse effects was different between the Drug & Placebo groups.

Here, the `proportions_ztest` function from the statsmodels package is useful. It has the following arguments.
- The number of relevant observations in each group; the `count` parameter. In this case, this quantifies the number of patients who experienced adverse effects in the Drug & Placebo groups.
- The number of total observations in each group; the `nobs` parameter. In this case, this quantifies the number of total patients in the Drug & Placebo groups.
- Use a 'two-sided' hypothesis test because the desire is to determine if there is a difference in general between the two groups, & not a difference in a specific direction.

```python
#Obtain the # individuals with adverse effects in each 'trx' group
num_adveffs_trx = orig_df[orig_df.adverse_effects == 'Yes'].trx.value_counts(dropna=False)

#Obtain the # individuals in each 'trx' group
#display(orig_df.trx.value_counts().values)

#Perform hypothesis test
z_score, two_sample_p_value = proportions_ztest(count=num_adveffs_trx.values, nobs=orig_df.trx.value_counts().values, 
                                     alternative='two-sided')
#Retrieve the coefficients
display(f"Z-score: {z_score}", f"P-value: {two_sample_p_value}")
```

The resulting z-score & p-value are 0.045 & 0.964 respectively. In hypothesis testing, the significance level, also known as alpha, is usually set no higher than 0.10. With the p-value being so much greater than this threshold, it is a clear indication that there is not sufficient evidence from which the null hypothesis can be rejected.  
> As such, there is not a significant difference in the proportion of individuals who experienced adverse effects from the drugs between the Drug & Placebo groups.

# Analysis II - Number of Adverse Effects Between Drug/Placebo Groups
**Find out if the number of adverse effects is independent of the treatment and control groups, saving as a variable called `num_effects_p_value` containing a p-value.**

This section will evaluate if there is a significant relationship between the number of adverse effects experienced by individuals & the treatment (drug) & control (placebo) groups. To start, a brief, exploratory analysis can be performed to get an initial evaluation of these variables.
- Keep in mind the number of individuals in each group: 10,727 in the treatment & 5,376 in the control (about a 2:1 ratio).

```python
#Obtain the # individuals per 'trx' & 'num_effects'
num_trx_numeff = orig_df.groupby(['trx','num_effects'])['age'].count().reset_index().rename(columns={'age':'#'})

#Want to add the proportions per 'trx' group
#calculate the proportions for the Drug, Placebo groups
prop_Drug_numeff = orig_df[orig_df.trx == 'Drug'][['trx','num_effects']].value_counts(normalize=True).reset_index()
prop_Plac_numeff = orig_df[orig_df.trx == 'Placebo'][['trx','num_effects']].value_counts(normalize=True).reset_index()
#merge the data, then append to the first dataset
trx_numeff_splits = num_trx_numeff.merge(pd.concat([prop_Drug_numeff, prop_Plac_numeff], axis=0), 
                                         on=['trx','num_effects']).rename(columns={'proportion':'% per trx group'})
#Change the new variable to a %
trx_numeff_splits['% per trx group'] = np.round(trx_numeff_splits['% per trx group']*100,2)

#Check the results
display(trx_numeff_splits.sort_values(['num_effects','trx']).reset_index(drop=True))
```

Since there are uneven quantities in the two groups (drug vs placebo), comparative analyses need to be performed. In other words, obtaining proportions across the 'trx' & 'num_effects' variables will be beneficial.

```python
#Illustrate the # adverse effects across 'trx'
ax1 = sns.barplot(data=trx_numeff_splits, x='num_effects', y='% per trx group', hue='trx')

#edit plot details
plt.title('Proportions of Patients per Group & the Number of Adverse Effects Experienced', fontsize=10)
plt.xlabel('# Adverse Effects Experienced')
plt.xticks(fontsize=9)
plt.ylabel('Proportion of Individuals per Group (%)', fontsize=9.5)
ax1.set(ylim=(0,100))
plt.yticks(ticks=range(0,101,10), fontsize=9)
plt.legend(title='Group', title_fontsize=9, fontsize=8, loc='upper right')
plt.grid(axis='y',alpha=0.2,color='gray')
plt.tight_layout();
```

The figure above illustrates the proportion of patients per group & the number of adverse effects they experienced from the drug usage.

Recall that in this trial, there were about two people in the "Drug" group for every one person in the "Placebo" group. Nonetheless, the proportions of individuals across the two treatment groups appear to be very similar. For example, the proportion of people who experienced one adverse effect was about 8.91% & 9.04% for those in the drug & placebo group respectively.  
As such, this figure suggests that the group that a patient was in did not have a significant effect on the number of adverse effects they experienced. Nevertheless, hypothesis testing can be utilized to evaluate this further.

This section is seeking to determine whether the group that a patient is in (the 'trx' variable) has an effect on the number of adverse effects they experience from the drug usage. In other words, are the 'trx' & 'num_effects' variables related. Such an inquiry can be measured with a chi-square test which is often used to evaluate whether two categorical variables are independent or not.

In this case, the null & alternative hypotheses follow as:
- $H_0$: The number of adverse effects experienced is independent of the group that the patient was in.
- $H_A$: The number of adverse effects experienced is _not_ independent of the group that the patient was in.

Here, the `chi2_independence` function from the pingouin package is useful. It has the following arguments.
- The `data` parameter takes the relevant dataset.
- The `x` & `y` parameters take the target & explanatory variables respectively.
- The `correction` parameter specifies whether or not to apply Yates' continuity correction, which is a fudge factor for when the sample size is very small & the degrees of freedom is 1. Note that degrees of freedom = (#-response-categories - 1) * (#-explanatory-categories).
    - Since each of the two groups has over 5,000 data points, applying this correction is unnecessary.
- The function returns 3 data frames: the expected counts, the observed counts, & statistics related to the test.

```python
#Perform the hypothesis test
expected, observed, stats = pingouin.chi2_independence(data=orig_df, x='num_effects', y='trx', correction=False)
display(stats)

#Assign the specified variable (in DataCamp)
num_effects_p_value = stats.loc[0,'pval']
#display(num_effects_p_value)
```

Even if a relatively large significance level of 0.10 is used for this hypothesis test of independence, the p-values in these results, the smallest of which is about 0.50, are still significantly larger. Therefore, there is not sufficient evidence from which the null hypothesis can be rejected.
> As such, the number of adverse effects experienced from the drug usage is independent of whether the patient was in the control (Placebo) or treatment (Drug) group.

The figures above demonstrates the distributions of ages of patients based on the trial group they were in; the Drug or Placebo group. Note that the vertical axes of the two plots are _not_ the same--that of the drug group is approximately twice as tall as that of the placebo group. This was done to help compare the overall shapes of the distributions.

Recall that in this trial, there were about two patients in the Drug group for every one patient in the Placebo group. The two groups have an average age of about 64 years. Generally, they appear to be quite similar in these plots since there are no major differences between them. As such, it is difficult to discern a significant difference in the ages of these patients across the two groups.

When performing hypothesis testing with numerical variables, evaluating their distributions can help to determine what kind of test is most appropriate. For example, parametric tests rely on the data being normally distributed, whereas non-parametric tests allow for exceptions to this.  
In this case, both of these two age distributions are close to, but not quite normal. There is a clear single peak near the midpoint of the age range, but neither of the distributions are symmetrical around the mean in the way that a true normal distribution would appear. Since the mean age is less than the mode of the age (the peak), the data are slightly left-skewed.  
As such, a non-parametric hypothesis test would be more appropriate with these variables.
- To evaluate the normality of these distributions further, other tests & visualizations could be used like a Q-Q plot.

Other factors to consider when choosing a hypothesis test include whether the data are independent or not, if there are clear outliers, or if the sample size is small. In this case, there are no signifcant outliers. Moreover, the two groups, which account for more than 5,000 observations each, are independent because every patient was only in one group & not both.  
Given this information, a Mann-Whitney U test could be appropriate for this data.

The `mwu` function from the pingouin package can be used to perform such a test. It has the following arguments.
- The `x` & `y` parameters take the numerical data from the two groups respectively.
- The `alternative` parameter specifies what kind of hypothesis test is being performed; one-tailed vs two-tailed. In this case, a two-tailed test is applicable because the objective pertains to whether there is a significant difference between the ages of the two groups.

Once more, take a significance value of 0.10. The null & alternative hypotheses are as follows.
- $H_0$: There is not a significant difference between the ages of the Drug & Placebo groups.
- $H_A$: There is a significant difference between the ages of the Drug & Placebo groups.

```python
#Perform the hypothesis test
mwu_test = pingouin.mwu(x=orig_df[orig_df.trx == 'Drug']['age'], y=orig_df[orig_df.trx == 'Placebo']['age'], 
                        alternative='two-sided')
#check the results
display(mwu_test)

#Retrieve the desired value (DataCamp)
age_group_effects_p_value = mwu_test['p-val'][0]
#display(age_group_effects_p_value)
```

This hypothesis test calculated a p-value of about 0.26, which is noticeably greater than the preestablished significance level of 0.10. Therefore, there is not sufficient evidence with which to reject the null hypothesis.
> As such, there is not a significant difference between the ages of the Drug & Placebo groups.
